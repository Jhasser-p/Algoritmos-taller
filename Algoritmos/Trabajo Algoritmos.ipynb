El problema de esta proyección es que es un modelo ideal; en el mundo real, la memoria del computador se puede llenar antes de terminar, o el código puede fallar por exceso de operaciones acumuladas.

Esto nos enseña que la notación teórica es como un modelo a gran escala que ignora la "fricción" interna: agrupa los algoritmos sin mirar los detalles finos del trabajo real que hace el procesador.
